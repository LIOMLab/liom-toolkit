## Save masks
Load the numpy mask files and save them to zarr

```python
import dask.array as da
import numpy as np
from liom_toolkit.utils import generate_label_color_dict_mask, load_zarr, save_label_to_zarr
from ome_zarr.dask_utils import resize
from tqdm.auto import tqdm
```

```python
# This notebook demonstrates attaching a mask label to an OME-Zarr store.
# The mask is a 3D .npy array (here a low-resolution brain mask extracted
# from the S23 zarr's mask label). We save it to a new small zarr for the
# demo; to attach to an existing volume, point zarr_files at your zarr.
import os
os.makedirs("data/vseg/output", exist_ok=True)
mask_files = ["/data/LSFM/S23/S23_mask.npy"]  # 128x128x128, 4MB
zarr_files = ["data/vseg/output/mask_demo.zarr"]
# Colour dict for setting the label colour in napari
colour_dict = generate_label_color_dict_mask()
scales = (6.5, 6.5, 6.5)
# The scale at which the mask is generated
resolution_level = 0
```

```python
for mask_file, zarr_file in tqdm(zip(mask_files, zarr_files)):
    mask = da.from_array(np.load(mask_file))
    # Switch x and z axis for zarr
    mask = da.transpose(mask, (2, 1, 0))

    save_label_to_zarr(mask, zarr_file, colour_dict, name="mask", scales=scales,
                       resolution_level=resolution_level)
print(f"Saved mask label to {zarr_files[0]}")
```
