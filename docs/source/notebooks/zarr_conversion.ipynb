## Lightsheet HDF5 to Zarr conversion

### Option 1: Single file to single zarr
This option will convert a single HDF5 file from the lightsheet microscope to a single zarr file. This is useful for small datasets that can fit into a single file or if you have single channel acquisitions.

```python
from liom_toolkit.conversion import convert_hdf5_to_zarr
```

```python
# Define the input & output paths and scales.
# This notebook uses real lab data on the LSFM server (/data/LSFM/...).
# These paths are not bundled — to re-run, point hdf5_file at your own
# HDF5 volume and zarr_file at an output path.
hdf5_file = "/data/LSFM/S23/555nm.hdf5"
zarr_file = "/data/LSFM/S23/S23_demo.zarr"
# The scales represent the physical resolution of the data in microns.
scales = (6.5, 6.5, 6.5)
```

```python
convert_hdf5_to_zarr(hdf5_file, zarr_file, scales=scales)
```

### Option 2: Full Zarr creation
This options use multichannel lightsheet acquisitions and will create a full zarr dataset with the following structure:
- 4D images of the acquisitions, with the dimensions (c, z, y, x)
- 3D mask (auto generated) of the acquisitions, with the dimensions (z, y, x)
- Allen CCF atlas aligned to the acquisition, with the dimensions (z, y, x)


```python
from liom_toolkit.conversion import create_full_zarr_volume
```

```python
auto_fluo_file = "/data/LSFM/S23/647nm.hdf5"
vascular_file = "/data/LSFM/S23/555nm.hdf5"
template_file = "/data/LSFM/template/p11/p11_lightsheet_25_15.nii"
# Path to an Allen CCF atlas NIfTI aligned to the template. Leave blank to
# use the auto-downloaded atlas (use_custom_atlas=False is the legacy path,
# see the known-issues notes in the docs).
atlas_file = ""
zarr_file = "/data/LSFM/S23/S23_full.zarr"
```

```python
# Create the full zarr volume (image + atlas + mask + vessels labels).
# This runs the full registration pipeline and is slow on real data; the
# basic HDF5->zarr conversion above is the lightweight demo. Wrapped so the
# notebook still renders if the full pipeline is skipped (e.g. no atlas file).
try:
    create_full_zarr_volume(auto_fluo_file, vascular_file, zarr_file,
                            template_path=template_file, atlas_path=atlas_file,
                            scales=scales)
except Exception as e:
    print(f"Full pipeline not run: {type(e).__name__}: {e}")
    print("The basic conversion (cell above) demonstrates HDF5 -> OME-Zarr.")
```
